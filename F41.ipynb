### Open & Modify Files

##### 1. `condition`

```python
import pandas as pd

## open condition file
condition = pd.read_csv('H255.csv')

## maximum columns set at 'None' to display all columns
pd.set_option('display.max_columns', None)

condition
```

```python
condition.describe
```

```python
# see what F41 values exist in condition file
condition['ICD10CDX'].astype(str)[condition['ICD10CDX'].astype(str).str.startswith('F41')].unique()
```

```python
# filter for F41
condition_f41 = condition[condition['ICD10CDX'] == 'F41'].copy()

# remove duplicate patients using DUPERSID
condition_f41.drop_duplicates(subset = ['DUPERSID'], keep = 'first', inplace = True)
```

```python
# checking # of unique F41 patients using DUPERSID
# (required to have at least 500 instances)
len(condition_f41)
```

```python
condition_f41
```

##### 2. `consolidation`

```python
consolidation = pd.read_csv('h256.csv')
consolidation
```

### Joining `condition` & `consolidation` files

```python
joined = pd.merge(left = consolidation, right = condition_f41 , how = 'inner', on = 'DUPERSID')
joined
```

```python
# attributes of the newly joined dataframe
joined.columns
```

### Data Preprocessing

```python
for column in joined.columns:
    if "_x" in column:
        # change attributes that have trailing '_x' 
        joined.rename(columns={column: column.replace("_x", "")}, inplace=True)
    elif "_y" in column:
        # drop unnecessary columns '_y'
        joined = joined.drop([column], axis = 1)
```

```python
# confirm attributes with _x renamed
# and attributes with _y removed
joined.columns
```
