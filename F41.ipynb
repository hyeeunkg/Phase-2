```python
## import pandas
import pandas as pd

## open file
data1 = pd.read_csv("H255.csv")

# test
data1.head()
```

```python
data1.describe
```

```python
data2= pd.read_csv('h256.csv')
data2.head(10)
```

```python
data = pd.merge(left = data2, right = data1 , how="inner", on='DUPERSID')
data
```

```python
data.columns
```

```python
## See how to use for loop

for column in data.columns:
    if "_x" in column:
        # change column names
        data.rename(columns={column: column.replace("_x", "")}, inplace=True)
    elif "_y" in column:
        # drop unnecessary columns column 
        data = data.drop([column], axis=1)
        
```
